```python
import sqlite3
```

```python
## conenct to sqlite 
connection=sqlite3.connect('Example.db')
connection
```

```python
cursor=connection.cursor()
```

```python
## craete a table
cursor.execute('''
    Create Table If Not Exists employees(
    id Integer Primary Key,
    name Text Not Null,
    age Integer,
    department text)
               ''')

## commit changes 
connection.commit()
```

```python
cursor.execute('''
Select * from employees
               ''')
```

```python
## insert the data in table

cursor.execute('''
    Insert into employees(name,age,department)
    VALUES('pratham','42','cse')
    ''')

cursor.execute('''
    Insert into employees(name,age,department)
    VALUES('atharv','32','cse')
    ''')

cursor.execute('''
    Insert into employees(name,age,department)
    VALUES('pravin','65','it')
    ''')

connection.commit()
```

```python
## quey the data from the data
cursor.execute(' select * from employees')
rows=cursor.fetchall()

## print the queried data

for row in rows:
    print(row)
```

```python
## update the data in the table
cursor.execute('''
    UPDATE employees
    Set age=100
    Where name="pratham"
               ''')

connection.commit()
```
